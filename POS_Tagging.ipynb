```python
import nltk
# Ensure necessary NLTK data is downloaded
nltk.download('punkt')
nltk.download('averaged_perceptron_tagger')
nltk.download('tagsets') # For help on tags
nltk.download('punkt_tab') # Download punkt_tab resource which is needed for tokenization
# Download the specific English averaged perceptron tagger
nltk.download('averaged_perceptron_tagger_eng')


sentences = [
    "The quick brown fox jumps over the lazy dog.",
    "A very beautiful car is parked outside.",
    "Natural language processing is fascinating.",
    "She quickly read the long book."
]

# Define patterns using specific tags (e.g., Penn Treebank)
# JJ: Adjective, NN: Noun (singular), NNS: Noun (plural)
# DT: Determiner, RB: Adverb, VB-any: Verb
patterns_to_find = {
    "Adjective-Noun": [('JJ', 'NN'), ('JJ', 'NNS')],
    "Determiner-Adjective-Noun": [('DT', 'JJ', 'NN'), ('DT', 'JJ', 'NNS')],
    "Adverb-Verb": [('RB', 'VB'), ('RB', 'VBD'), ('RB', 'VBG'), ('RB', 'VBN'), ('RB', 'VBP'), ('RB', 'VBZ')]
}

def find_pos_patterns(text, patterns):
    tokens = nltk.word_tokenize(text)
    tagged_tokens = nltk.pos_tag(tokens)
    print(f"\nSentence: {text}")
    print(f"Tagged: {tagged_tokens}")

    found_patterns_in_sentence = {}

    for pattern_name, pattern_list_of_tuples in patterns.items():
        for pattern_tuple in pattern_list_of_tuples: # e.g., pattern_tuple is ('JJ', 'NN')
            pattern_len = len(pattern_tuple)
            for i in range(len(tagged_tokens) - pattern_len + 1):
                # Extract the tags of the current window
                current_window_tags = tuple(tagged_tokens[j][1] for j in range(i, i + pattern_len))

                if current_window_tags == pattern_tuple:
                    # Extract the words of the current window
                    matched_words = ' '.join(tagged_tokens[j][0] for j in range(i, i + pattern_len))
                    if pattern_name not in found_patterns_in_sentence:
                        found_patterns_in_sentence[pattern_name] = []
                    found_patterns_in_sentence[pattern_name].append(matched_words)

    if found_patterns_in_sentence:
        print("Found Patterns:")
        for name, matches in found_patterns_in_sentence.items():
            print(f"  {name}: {', '.join(matches)}")
    else:
        print("No defined patterns found.")

# Process each sentence
for sentence in sentences:
    find_pos_patterns(sentence, patterns_to_find)

# To see tagset help:
# nltk.help.upenn_tagset('RB')
# nltk.help.upenn_tagset('JJ')
# nltk.help.upenn_tagset('NN')
```
